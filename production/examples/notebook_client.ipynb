# Labor-cost analysis

Select a published model and dataset. The installed library performs all calculations. No payroll data or results are stored in this notebook.
Install the reporting extra (included in `[notebook]`). Charts reuse stored results; after analysis, rerun only the report cell to change presentation.


```python
from pathlib import Path

from labor_cost import (
    AnalysisRequest,
    Catalog,
    analyze,
    export_run,
    render_model,
    report,
)

catalog_path = None  # Path to your published catalog.json
model_id = "labor-cost"
model_version = "1.0.0"
dataset_id = "payroll"
dataset_version = None
period_a = None
period_b = None
selected_teams = []
selected_driver = None  # Leaf ID from the report driver table, not a label.
currency_label = "Currency"
report_renderer = "plotly_mimetype"  # JupyterLab/VS Code; classic: "notebook"
source_cost_column = None
explainer = "exact"
run_directory = None
```

```python
model = None
if catalog_path is not None:
    catalog = Catalog(Path(catalog_path))
    model = catalog.model(model_id, model_version)
    display(render_model(model))
else:
    print("Set catalog_path to preview a published model.")
```

```python
result = None
if all(
    x is not None for x in (catalog_path, dataset_version, period_a, period_b)
):
    dataset = catalog.dataset(dataset_id, dataset_version)
    request = AnalysisRequest(
        period_a,
        period_b,
        source_cost_column=source_cost_column,
        explainer=explainer,
    )
    result = analyze(model, dataset, request)
    print("Validated run:", result.run_id)
else:
    print("Set catalog, dataset version and comparison periods above.")
```

```python
# Rerun this cell to change charts; attribution is not rerun.
charts = None
if result is not None:
    charts = report(
        result, teams=selected_teams, driver=selected_driver,
        currency_label=currency_label, renderer=report_renderer,
    )
    if run_directory is not None:
        print("Run bundle:", export_run(result, run_directory))
```
