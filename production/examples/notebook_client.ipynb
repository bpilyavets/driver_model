# Labor-cost analysis

Select a published model and dataset. The installed library performs all calculations. No payroll data or results are stored in this notebook.

```python
from pathlib import Path

from labor_cost import AnalysisRequest, Catalog, analyze, export_run

catalog_path = None  # Path to your published catalog.json
model_id = "labor-cost"
model_version = "1.0.0"
dataset_id = "payroll"
dataset_version = None
period_a = None
period_b = None
selected_teams = []
source_cost_column = None
explainer = "exact"
run_directory = None
```

```python
result = None
if all(
    x is not None for x in (catalog_path, dataset_version, period_a, period_b)
):
    catalog = Catalog(Path(catalog_path))
    model = catalog.model(model_id, model_version)
    dataset = catalog.dataset(dataset_id, dataset_version)
    request = AnalysisRequest(
        period_a,
        period_b,
        source_cost_column=source_cost_column,
        explainer=explainer,
    )
    result = analyze(model, dataset, request)
    print("Validated run:", result.run_id)
else:
    print("Set catalog, dataset version and comparison periods above.")
```

```python
if result is not None:
    display(result.table("validation"))
    if selected_teams:
        view = result.select_teams(selected_teams)
        display(view.team_summary)
    if run_directory is not None:
        print("Dashboard bundle:", export_run(result, run_directory))
```
